## Scrum Dynamic Nested Sequential Approach

### Definition:
- Dynamic sprint execution with nested feedback loops through Scrum Master
- Scrum Master coordinates with each team member through explicit gates
- Each team member completes work and reports back to Scrum Master
- Scrum Master reviews, relates to previous work, and approves before next step
- **Key Feature**: Sprint count determined DYNAMICALLY by remaining Product Backlog
- Sprint count is NOT hardcoded - it adapts based on velocity and remaining work
- Scrum Master signals completion when all high-priority stories are done

### Dynamic Sprint Pattern:
```python
While (high-priority stories remain AND max_turns not reached):
    Sprint N:
        product_owner ↔ scrum_master (planning/review)
        scrum_master ↔ developer (development coordination)
        developer → scrum_master (completion feedback)
        scrum_master ↔ qa_engineer (testing coordination)
        qa_engineer → scrum_master (quality feedback)
        scrum_master ↔ designer (design coordination)
        scrum_master ↔ devops (infrastructure coordination)
        scrum_master ↔ product_owner (review/retrospective)
        scrum_master → Check backlog status → Decide: Continue OR Complete
```

### Key Feature:
Scrum Master acts as central coordinator with explicit feedback after each ceremony/phase.
Number of sprints is determined by actual backlog completion, not predetermined.

**Key Difference from Hardcoded Approach**:
- Dynamic: Sprint count determined by actual backlog completion (backlog-driven)
- Hardcoded: Fixed 3-sprint structure (predetermined)

This approach maintains SM-coordinated workflow while adapting sprint count based on velocity and remaining work.

## Agent Class Definition

````python
class Agent:
    """Custom Agent class for Scrum multi-agent conversations"""
    
    def __init__(self, name, system_prompt, llm):
        self.name = name
        self.llm = llm
        self.system_prompt = system_prompt
        
        # Create chat prompt with conversation history
        from langchain_core.messages import SystemMessage
        self.chat_prompt = ChatPromptTemplate.from_messages([
            SystemMessage(content=self.system_prompt),
            HumanMessagePromptTemplate.from_template("{conversation_history}\n{input}")
        ])
        
        # Create chain using LCEL (LangChain Expression Language)
        self.chain = self.chat_prompt | self.llm
    
    def respond(self, input_text, conversation_history):
        """Generate response based on input and conversation history"""
        # Invoke the chain with the input
        response = self.chain.invoke({
            "input": input_text,
            "conversation_history": conversation_history
        })
        return response.content

print("✅ Agent class defined")
````

## Create All Scrum Agents

````python
# Product Owner Agent - WITH DYNAMIC BACKLOG TRACKING
product_owner_agent = Agent(
    name="Product_Owner",
    system_prompt="""You are the Product Owner for Electric Appliances E-Commerce Platform.
    
    PRODUCT BACKLOG MANAGEMENT (DYNAMIC):
    First, create the Product Backlog with user stories in the format: 'As a [user type], I want [feature], so that [benefit]'.
    Assign each user story a unique ID (US-001, US-002, etc.), priority (High/Medium/Low), and business value (1-10).
    Create 20-25 user stories total.
    
    DYNAMIC SPRINT PLANNING:
    During each Sprint Planning:
    - Review remaining backlog (track completed vs remaining)
    - Select high-priority stories based on team velocity
    - Report status: "Total: X stories | Completed: Y | High-priority remaining: Z"
    
    During each Sprint Review:
    - Accept/reject completed stories
    - Update backlog status  
    - Work with Scrum Master to decide: Continue to next sprint OR Project complete?
    - Signal completion when: All high-priority stories done OR MVP delivered
    
    Then calculate the {effort} to create the Product Backlog in hours, days, weeks, or months, follow the steps below:
        - step 1. work = {the estimate of total number of user stories created and prioritized for the product backlog}
        - step 2. productivity rate = {5-8 user stories documented every day}
        - step 3. effort = {work}/{productivity}
    
    The amount of {work} in this phase is measured by the total number of user stories created and prioritized in the Product Backlog.
    """,
    llm=llm
)

# Scrum Master Agent - DYNAMIC SPRINT MANAGEMENT (Central Coordinator)
scrum_master_agent = Agent(
    name="Scrum_Master",
    system_prompt="""You are the Scrum Master for Electric Appliances E-Commerce Platform.
    
    DYNAMIC SPRINT APPROACH - NO FIXED SPRINT COUNT:
    Your role is to facilitate Scrum ceremonies and track sprints DYNAMICALLY based on remaining work.
    
    Sprint Tracking:
    - Start with Sprint 0 (backlog creation + estimation)
    - Then Sprint 1, 2, 3... continue until work is done
    - DO NOT assume a fixed number of sprints
    - Let the actual backlog drive how many sprints are needed
    
    After Each Sprint Retrospective:
    1. Review remaining Product Backlog with Product Owner
    2. Check: Are there high-priority stories remaining?
    3. Decide: "Continue to next sprint" OR "Project complete"
    4. If continuing: Announce "Planning Sprint X..." (where X is next number)
    5. If complete: Say "SCRUM_PROJECT_COMPLETE - All priority work delivered. Total sprints: X"
    
    First, work with the team to estimate story points for user stories using Planning Poker (Fibonacci: 1, 2, 3, 5, 8, 13).
    Help select user stories for Sprint Backlog based on team velocity (typically 20-30 story points per 2-week sprint).
    
    Then calculate the {effort} for facilitating sprint ceremonies in hours, days, weeks, or months, follow the steps below:
        - step 1. work = {number of sprint ceremonies facilitated: Sprint Planning, Daily Scrums (10 per sprint), Sprint Review, Sprint Retrospective}
        - step 2. productivity rate = {facilitate 4 major ceremonies per sprint + 10 daily scrums = 14 ceremonies per 2-week sprint}
        - step 3. effort = {work}/{productivity}
        - step 4. velocity tracking = {story points completed / sprint duration}
    
    The amount of {work} is measured by number of ceremonies facilitated and velocity metrics tracked.
    
    CRITICAL - WHEN TO SIGNAL COMPLETION:
    Only signal completion when:
    - All high-priority user stories are completed, OR
    - Product Owner declares MVP is delivered, OR
    - Remaining backlog contains only low-priority items and PO approves stopping
    
    THEN say: 'SCRUM_PROJECT_COMPLETE - All priority work delivered. Completed X sprints with Y story points total.'
    """,
    llm=llm
)

# Developer Agent - WITH DYNAMIC PARTICIPATION
developer_agent = Agent(
    name="Developer",
    system_prompt="""You are a Senior Developer for Electric Appliances E-Commerce Platform.
    
    First, implement user stories from the Sprint Backlog by writing source code based on the Product Backlog and acceptance criteria.
    Estimate implementation in story points (1=~4hrs, 3=~1day, 5=~2days, 8=~3days, 13=too large, break down).
    
    During Sprint Planning:
    - Help estimate remaining stories in backlog
    - Provide technical input on story complexity
    - Commit to realistic sprint capacity based on velocity
    
    Then calculate the {effort} to implement user stories in hours, days, weeks, or months, follow the steps below:
        - step 1. work = {total story points committed in the sprint backlog}
        - step 2. productivity rate = {team velocity: typically 20-30 story points completed per 2-week sprint}
        - step 3. effort = {work}/{productivity} = {number of sprints needed}
        - step 4. SLOC estimate = {story points × 100-200 lines of code per point}
    
    The amount of {work} is measured by total story points in sprint backlog and corresponding SLOC.
    Story points: 1=~4hrs, 3=~1day, 5=~2days, 8=~3days
    
    In nested sequential discussions:
    - Share implementation approach and estimates with Scrum Master
    - Report development progress through SM coordination
    - Discuss technical challenges in feedback loops
    - Collaborate with QA and Designer via Scrum Master
    """,
    llm=llm
)

# QA Engineer Agent - WITH WORK/PRODUCTIVITY/EFFORT FORMULAS
qa_engineer_agent = Agent(
    name="QA_Engineer",
    system_prompt="""You are a QA Engineer for Electric Appliances E-Commerce Platform.
    
    First, create test cases for user stories in the Sprint Backlog and test during the sprint (not at the end).
    Verify acceptance criteria are met for each user story before it can be marked as Done.
    
    Then calculate the {effort} to create and execute test cases in hours, days, weeks, or months, follow the steps below:
        - step 1. work = {total number of test cases created based on story points in sprint backlog}
        - step 2. productivity rate = {2-3 test cases created and executed per story point}
        - step 3. effort = {work}/{productivity}
        - step 4. test cases estimate = {story points × 2.5 test cases per point}
    
    The amount of {work} is measured by total test cases created and executed during the sprint.
    Testing: ~2-3 test cases per story point
    
    In nested sequential discussions:
    - Share testing strategy and progress with Scrum Master
    - Report bugs and quality metrics through feedback loops
    - Collaborate on acceptance criteria via SM coordination
    - Provide quality feedback in handoff conversations
    """,
    llm=llm
)

# UI/UX Designer Agent - WITH WORK/PRODUCTIVITY/EFFORT FORMULAS
designer_agent = Agent(
    name="UX_Designer",
    system_prompt="""You are a UI/UX Designer for Electric Appliances E-Commerce Platform.
    
    First, create wireframes, mockups, and design deliverables for user stories in the Sprint Backlog.
    Ensure mobile-responsive, consistent design system across all features.
    
    Then calculate the {effort} to create design deliverables in hours, days, weeks, or months, follow the steps below:
        - step 1. work = {total number of screens/mockups/wireframes to be designed based on user stories}
        - step 2. productivity rate = {3-5 screens designed per day}
        - step 3. effort = {work}/{productivity}
        - step 4. design estimate = {estimate screens needed per user story, typically 1-3 screens per story}
    
    The amount of {work} is measured by total design deliverables (screens, wireframes, mockups) created.
    
    In nested sequential discussions:
    - Share design approach and deliverables with Scrum Master
    - Discuss UX considerations in feedback loops
    - Collaborate with developers via SM coordination
    - Provide design feedback through handoff conversations
    """,
    llm=llm
)

# DevOps Engineer Agent - WITH WORK/PRODUCTIVITY/EFFORT FORMULAS
devops_agent = Agent(
    name="DevOps_Engineer",
    system_prompt="""You are a DevOps Engineer for Electric Appliances E-Commerce Platform.
    
    First, set up CI/CD pipelines, infrastructure, deployment automation, and monitoring for the sprint increment.
    Ensure stable demo environment for Sprint Review and production deployment readiness.
    
    Then calculate the {effort} for DevOps tasks in hours, days, weeks, or months, follow the steps below:
        - step 1. work = {number of infrastructure tasks: CI/CD setup, environment config, deployment automation, monitoring setup}
        - step 2. productivity rate = {1-2 infrastructure tasks completed per day}
        - step 3. effort = {work}/{productivity}
        - step 4. estimate tasks per sprint = {typically 3-5 DevOps tasks per sprint}
    
    The amount of {work} is measured by number of infrastructure and DevOps tasks completed.

    In nested sequential discussions:
    - Share infrastructure status with Scrum Master
    - Report on deployment and monitoring through feedback loops
    - Discuss DevOps requirements via SM coordination
    - Collaborate on technical solutions in handoff conversations
    """,
    llm=llm
)

print(f"{GREEN}✅ All 6 Scrum agents defined with DYNAMIC sprint management!{RESET}")
````

## Define Agent Order for Dynamic Nested Sequential

### Nested Sequential with Scrum Master Coordination:
- Scrum Master acts as central coordinator
- Each agent interacts with SM through explicit feedback loops
- Dynamic sprint count based on remaining backlog
- SM tracks progress and signals completion when appropriate

````python
# Define agent order for DYNAMIC nested sequential with SM coordination
# SM coordinates with each team member in round-robin order
agents_order = [
    product_owner_agent,
    scrum_master_agent,
    developer_agent,
    qa_engineer_agent,
    designer_agent,
    devops_agent
]

print(f"{GREEN}✅ Dynamic nested sequential agent order defined with {len(agents_order)} agents{RESET}")
print(f"{GREEN}   Scrum Master acts as central coordinator for all interactions{RESET}")
````

## Execute Dynamic Scrum Nested Sequential Workflow

````python
print(f"{BLUE_BOLD}{'='*80}{RESET}")
print(f"{BLUE_BOLD}STARTING DYNAMIC SCRUM NESTED SEQUENTIAL EXECUTION{RESET}")
print(f"{BLUE_BOLD}Electric Appliances E-Commerce Platform{RESET}")
print(f"{BLUE_BOLD}Phase 2: Scrum Agile with SM Coordination{RESET}")
print(f"{BLUE_BOLD}{'='*80}{RESET}\n")

# DYNAMIC Configuration - No fixed sprint count
max_turns = 40  # Safety limit (4 agents × 10 rounds = ~5-6 sprints)
conversation_history = ""
project_complete = False

print(f"{GREEN}Dynamic Scrum Nested Sequential Configuration:{RESET}")
print(f"  - Agents: {len(agents_order)}")
print(f"  - Sprint Count: DYNAMIC (backlog-driven)")
print(f"  - Max Turns: {max_turns} (safety limit)")
print(f"  - Termination: When Scrum Master signals PROJECT_COMPLETE")
print(f"  - Pattern: SM-coordinated feedback loops until all priority work done\n")
print(f"{'='*80}\n")

# Initial customer message
print(f"{BLUE_BOLD}Customer (Initial Request):{RESET}\n")
print(f"{customer_message}\n")
print("-" * 80)
conversation_history += f"Customer: {customer_message}\n\n"

# DYNAMIC Nested Sequential: Continue until project complete or max turns reached
turn = 0
current_round = 1

while turn < max_turns and not project_complete:
    # Calculate current agent (round-robin through all agents)
    agent_idx = turn % len(agents_order)
    current_agent = agents_order[agent_idx]
    
    # Check for round completion
    if turn > 0 and turn % len(agents_order) == 0:
        current_round += 1
        print(f"\n{GREEN}{'='*80}{RESET}")
        print(f"{GREEN}✓ ROUND {current_round - 1} COMPLETED - All {len(agents_order)} agents contributed{RESET}")
        print(f"{GREEN}{'='*80}{RESET}\n")
    
    print(f"\n{CYAN}[Round {current_round}, Turn {turn + 1}]{RESET}")
    print(f"{BLUE_BOLD}Speaker: {current_agent.name}{RESET}")
    print("-" * 80)
    
    # Create context prompt based on turn and role
    if turn == 0:
        # First agent (Product Owner) in first round
        prompt = f"""You are the Product Owner. The customer has requested:

{customer_message}

Begin creating the Product Backlog with user stories for this Scrum project.
Coordinate with the Scrum Master."""
    else:
        # All other turns - use conversation history with role-specific guidance
        if current_agent.name == "Scrum_Master":
            prompt = f"""As Scrum Master (central coordinator), review the team's progress and provide your facilitation and coordination.

Recent conversation:
{conversation_history[-2500:]}

Coordinate the next step in the Scrum process based on where we are."""
        else:
            prompt = f"""Based on the Scrum Master's coordination and team discussion, provide your input for your role.

Recent conversation:
{conversation_history[-2500:]}

Provide your analysis, deliverables, or feedback. Coordinate back with the Scrum Master."""
    
    # Agent responds
    response = current_agent.respond(prompt, conversation_history)
    
    # Update conversation history
    conversation_history += f"{current_agent.name}: {response}\n\n"
    
    # Display
    print(f"\n{current_agent.name}:\n")
    print(response)
    print("\n" + "=" * 80)
    
    # Check for completion signal from Scrum Master
    if "PROJECT_COMPLETE" in response and current_agent.name == "Scrum_Master":
        project_complete = True
        print(f"\n{GREEN}{'='*80}{RESET}")
        print(f"{GREEN}✓ PROJECT COMPLETION SIGNAL RECEIVED FROM SCRUM MASTER{RESET}")
        print(f"{GREEN}{'='*80}{RESET}\n")
        break
    
    turn += 1

# Final summary
print(f"\n{GREEN}{'='*80}{RESET}")
print(f"{GREEN}✓ DYNAMIC SCRUM NESTED SEQUENTIAL EXECUTION COMPLETE{RESET}")
print(f"{GREEN}{'='*80}{RESET}")
print(f"\nTotal turns: {turn}")
print(f"Rounds completed: {current_round}")
print(f"Completion reason: {'Scrum Master signaled project complete' if project_complete else 'Max turns reached'}")
print(f"Sprint count: Dynamic (based on actual backlog completion with SM coordination)")
````
